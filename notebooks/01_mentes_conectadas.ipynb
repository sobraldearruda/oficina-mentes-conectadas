## 2. 🧩 Engenharia de features

**Features** (características) são as informações que descrevemos sobre algo para que uma IA consiga observar, comparar e decidir. É como se a IA só enxergasse o mundo através das características que damos a ela.

### Exemplo guiado: maçã ou banana?

Vamos montar, com a turma, uma pequena tabela de features que ajudariam uma IA a diferenciar essas duas frutas.


```python
# Tabela de features: cada linha é uma fruta, cada coluna é uma feature (característica)
import pandas as pd

frutas = pd.DataFrame({
    "fruta": ["maçã", "banana"],
    "cor": ["vermelha", "amarela"],
    "formato": ["arredondado", "alongado"],
    "tamanho_cm": [8, 18],
})

frutas
```

Agora vamos criar uma função bem simples que usa essas features para "decidir" qual fruta é qual — um exemplo minúsculo (e manual) do que uma IA faz em grande escala, automaticamente, a partir de muitos dados.

```python
# Uma função simples que decide a fruta usando as features "cor" e "formato"
def qual_fruta(cor, formato):
    if cor == "amarela" and formato == "alongado":
        return "🍌 Banana"
    elif cor in ("vermelha", "verde") and formato == "arredondado":
        return "🍎 Maçã"
    else:
        return "🤔 Não sei — features insuficientes ou não reconhecidas"

# Testando a função com algumas combinações de features
print(qual_fruta("amarela", "alongado"))
print(qual_fruta("vermelha", "arredondado"))
print(qual_fruta("verde", "arredondado"))
```

**Para explorar com a turma:** o que acontece se testarmos uma combinação de features que a função não conhece, como `qual_fruta("roxa", "alongado")`? Use a célula abaixo para testar — isso ajuda a discutir por que as IAs reais precisam de **muitos exemplos e muitas features** para aprender padrões mais gerais, em vez de regras fixas como essa.

```python
# ✏️ Teste aqui outras combinações de cor e formato
qual_fruta("roxa", "alongado")
```
